##Variáveis Qualitativas

São aquelas que não podem ser medidas numericamente, mas sim em termos de qualidade ou categoria. Elas são divididas em dois tipo:

***Nominais:*** Não possuem uma ordem natural entre as categorias. Exemplo: Cargo

***Ordinais:*** Possuem uma ordem ou classificação. Exemplos: nivel de experiencia ( Júnior, Pleno, Senior)

## **Variávies Quantitativas**

São aquelas que podem ser medidas numericamente e expressa quantidade e magnetude. E também podem ser classificadas em dois tipos.

Contínuas: Assume qualquer valor dentro de um intervalo. Exemplo: altura

Discretas: Assumem qualquer valor Inteiro. Exemplo: Ano do dado informado, Salário, etc

## **Valores Nulos (Missing Values)**

**O que são?**

Valores nulos, frequentemente representados como `NaN` (Not a Number) ou `None` em Python/Pandas, são entradas ausentes ou indefinidas em um conjunto de dados. Eles indicam que não há informação disponível para uma determinada observação em uma coluna específica.

**Por que devem ser analisados?**

  Integridade dos Dados, Precisão da Análise, Viés e Conclusões Falsas, Desempenho de Modelos, Tomada de Decisão

## **Dados Duplicados (Duplicate Data)**

**O que são?**

Dados duplicados são registros ou observações idênticas (ou quase idênticas) que aparecem mais de uma vez em um conjunto de dados. Isso pode ocorrer em todas as colunas ou em um subconjunto de colunas que, em conjunto, identificam um registro único.

**Por que devem ser analisados?**

A remoção ou tratamento de dados duplicados é fundamental por algumas razões:

  Exatidão dos Dados Eficiência, Consistência, Análise e Modelagem, Qualidade dos Dados,

```python
df.T.duplicated()
```

##Como valores nulos podem afetar análises e conclusões?
Em análises estatísticas, a presença de valores nulos pode distorcer os resultados e levar a conclusões errôneas. Isso ocorre porque a maioria dos algoritmos estatísticos não é feita para lidar com valores nulos, o que pode resultar em imprecisão nos resultados.

##Em que situações informações duplicadas podem gerar problemas

Informações duplicadas podem gerar problemas significativos, tais como:

**`Distorção de Análises:`** Levam a contagens, médias e somas incorretas, falseando a realidade dos dados.

**`Viés em Modelos de Machine Learning:`** Podem causar overfitting e fazer com que modelos aprendam padrões irreais.

**`Ineficiência:`** Consomem espaço desnecessário e diminuem a velocidade de processamento e consultas.

**`Inconsistência de Dados:`** Criam conflitos de informação se nem todos os registros duplicados forem atualizados de forma uniforme.

**`Decisões de Negócio Falhas:`** Podem levar a escolhas equivocadas devido a relatórios e análises imprecisas.

##**Quais estatísticas são calculadas?**

Ao utilizar um método como o .describe() em um DataFrame (especialmente para variáveis numéricas), as estatísticas mais comuns calculadas são:

count (contagem), mean (média), std (desvio padrão), min (mínimo), 25 (primeiro quartil), 50% (mediana ou segundo quartil), 75% (terceiro quartil),
max (máximo)

##**Para quais tipos de variáveis esse método se aplica?**

Por padrão, o método describe() é aplicado a variáveis quantitativas (numéricas) e retorna estatísticas descritivas como contagem, média, desvio padrão, mínimos, máximos e quartis.
Já para variáveis qualitativas (categóricas), é preciso especificar isso (por exemplo, usando include='object' ou include='all'). Nesses casos, o describe() retorna principalmente: contagem, número de valores únicos (unique), valor mais frequente (top) e a frequência desse valor (freq).

# **Graficos:**



##**Grafico de Barra**

**Relação com os Dados:** Ideais para comparar quantidades discretas ou categorias. Cada barra representa uma categoria, e o comprimento da barra indica seu valor. Podem ser verticais ou horizontais.

**Quando usar:** Comparar vendas de produtos diferentes, número de alunos por curso, contagem de observações por categoria (como fizemos com categoria_cargo). Ótimos para rankings.

##**Grafico de linhas**

**Relação com os Dados:** Perfeitos para exibir tendências ao longo do tempo. O eixo X geralmente representa o tempo (datas, anos) e o eixo Y, os valores correspondentes.

**Quando usar:** Mostrar a evolução de um salário ao longo dos anos, o crescimento de vendas mensais, a variação da temperatura diária. Útil para identificar padrões, picos e vales.

##**Grafico de pizza**
**Relação com os Dados:** Usados para mostrar a proporção de cada categoria em relação ao total (o "todo"). Cada fatia representa uma categoria, e o tamanho da fatia corresponde à sua porcentagem do total.

**Quando usar:** Ideal para visualizar a composição de um todo. Exemplos: fatia de mercado por empresa, distribuição percentual de respostas em uma pesquisa. Cuidado: Não são recomendados para muitas categorias, pois se tornam ilegíveis.

##**Histograma**
**Relação com os Dados:** Representam a distribuição de frequência de uma única variável quantitativa. Dividem os dados em "caixas" (bins) e mostram quantas observações caem em cada caixa.

**Quando usar:** Entender a forma, o centro e a dispersão dos dados. Identificar se a distribuição é normal, assimétrica, multimodal, etc. Exemplos: distribuição da idade dos funcionários, distribuição dos salario_dolar.

##**Boxplot**
**Relação com os Dados:** Mostram a distribuição de uma variável quantitativa, especialmente útil para comparar distribuições entre diferentes grupos (categorias). Exibem a mediana, os quartis (25% e 75%), e identificam outliers.

**Quando usar:** Comparar a distribuição de salários entre diferentes experience_level, ou a variação de notas por turma. Ótimos para identificar assimetrias e outliers de forma concisa.

##**Grafico de dispersão**

**Relação com os Dados:** Mostram a relação entre duas variáveis quantitativas. Cada ponto no gráfico representa uma observação, com sua posição determinada pelos valores das duas variáveis nos eixos X e Y.

**Quando usar:** Identificar correlações, padrões ou agrupamentos entre variáveis. Exemplos: relação entre salario_dolar e work_year, ou entre horas trabalhadas e produtividade. Útil para detectar outliers.

```python
edia_salario_por_ano = df_pt_br.groupby(df_pt_br['ano_trabalho'].dt.year)['salario_dolar'].mean().sort_values()
percentage_increase = media_salario_por_ano.pct_change() * 100

jobs_ml_ia_por_ano = df_ml_ia['ano_trabalho'].dt.year.value_counts().sort_index()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))


sns.barplot(
    x=media_salario_por_ano.index,
    y=media_salario_por_ano.values,
    palette='viridis',
    ax=axes[0]
)

axes[0].set_title('Média Salarial Anual')
axes[0].set_xlabel('Ano')
axes[0].set_ylabel('Salário (USD)')
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Anotações %
for i, (year, value) in enumerate(media_salario_por_ano.items()):
    if not pd.isna(percentage_increase[year]):
        axes[0].text(
            i,
            value + (media_salario_por_ano.max() * 0.01),
            f'+{percentage_increase[year]:.1f}%',
            ha='center',
            fontsize=9
        )

axes[0].tick_params(axis='x', rotation=45)


sns.lineplot(
    x=jobs_ml_ia_por_ano.index,
    y=jobs_ml_ia_por_ano.values,
    marker='o',
    color='purple',
    ax=axes[1]
)

axes[1].set_title('Crescimento de Cargos ML/IA')
axes[1].set_xlabel('Ano')
axes[1].set_ylabel('Número de Cargos')
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

axes[1].set_xticks(jobs_ml_ia_por_ano.index)


plt.tight_layout()
plt.show()
```
